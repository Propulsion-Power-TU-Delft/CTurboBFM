```python
import numpy as np
import pandas as pd
import pickle
import os
import matplotlib.pyplot as plt
import pandas as pd
from unsflow.utils.thesis_plots import *
```

```python
su2Files = [
    "../su2_roe1/su2_lower_wall.csv",
    "../su2_jst/su2_lower_wall.csv"
    ]

ctbfmFiles = [
    "../ctbfm_roe1/Volume_CSV/results.csv",
    "../ctbfm_jst/Volume_CSV/results.csv",
    ]

labels = [
    'Roe',
    'JST',
    ]

ls = [
    '-',
    '--',
    ]
```

```python
su2s = []
for su2File in su2Files:
    su2s.append(pd.read_csv(su2File))
```

```python
ctbfms = []
for ctbfmFile in ctbfmFiles:
    with open(ctbfmFile, 'r') as f:
        ni = int(f.readline().strip().split('=')[1])
        nj = int(f.readline().strip().split('=')[1])
        nk = int(f.readline().strip().split('=')[1])

    df = pd.read_csv(ctbfmFile, skiprows=3)
    ctbfm = {col: df[col].to_numpy().reshape((ni, nj, nk)) for col in df.columns}
    ctbfms.append(ctbfm)
```

```python

# Create figure and subplots (1 row, 2 columns)
set_thesis_style()
fig, (ax1, ax2) = create_figure(fraction=1.0, aspect_ratio=1.45, subplots=(1, 2))

# ========== Left subplot: Mach number ==========
for ii, su2 in enumerate(su2s):
    ax1.plot(su2['Points_0'], su2['Mach'], '-C%i' %ii, mfc='none', label='SU2 - %s' % labels[ii])
    ax2.plot(su2['Points_0'], su2['Pressure'] / 1e5, '-C%i' %ii, mfc='none')

for ii, ctbfm in enumerate(ctbfms):
    ax1.plot(ctbfm['x'][:, 0, 0], ctbfm['Mach'][:, 0, 0], '--C%i' %(ii+2), mfc='none', label='Current - %s' % labels[ii])
    ax2.plot(ctbfm['x'][:, 0, 0], ctbfm['Pressure'][:, 0, 0] / 1e5, '--C%i' %(ii+2), mfc='none', label=labels[ii])

ax1.set_xlabel(r'$x$ [m]')
ax1.set_ylabel(r'$M$')
ax1.grid(alpha=0.2)
ax1.legend()
ax2.set_xlabel(r'$x$ [m]')
ax2.set_ylabel(r'$p$ [bar]')
ax2.grid(alpha=0.2)
lims = [0.35, 0.85]
ax1.set_xlim(lims)
ax2.set_xlim(lims)

# Save figure
plt.savefig("SupersonicWedge_Mach_Pressure_subplots.pdf")

# plt.show()
```
